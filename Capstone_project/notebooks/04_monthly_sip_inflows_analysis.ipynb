```python
df_04['year'] = df_04['month'].dt.year

# Calculate total and average monthly SIP inflow per year
yearly_sip = df_04.groupby('year').agg(
    total_sip_inflow=('sip_inflow_crore', 'sum'),
    avg_monthly_inflow=('sip_inflow_crore', 'mean')
).reset_index()

# Calculate year-over-year growth
yearly_sip['growth_pct'] = (
    yearly_sip['total_sip_inflow'].pct_change() * 100
)

yearly_sip
```

```python
fig = px.bar(
    yearly_sip,
    x='year',
    y='total_sip_inflow',
    text='total_sip_inflow',
    title='Year-wise Total SIP Inflows (2022–2025)',
    labels={
        'year': 'Year',
        'total_sip_inflow': 'Total SIP Inflow (₹ Crore)'
    }
)

fig.update_traces(
    texttemplate='₹%{y:,.0f} Cr',
    textposition='outside'
)

fig.update_layout(
    template='plotly_white',
    width=1100,
    height=600
)

fig.show()
```

```python
fig.write_image("../reports/plots/yearly_sip_inflow.png")
```

```python
fig = px.bar(
    yearly_sip.dropna(subset=['growth_pct']),
    x='year',
    y='growth_pct',
    text='growth_pct',
    title='Year-over-Year SIP Inflow Growth (2023–2025)',
    labels={
        'year': 'Year',
        'growth_pct': 'Growth (%)'
    }
)

fig.update_traces(
    texttemplate='%{y:.2f}%',
    textposition='outside'
)

fig.update_layout(
    template='plotly_white',
    width=1100,
    height=600,
    yaxis_title='YoY Growth (%)'
)

fig.show()
```

```python
fig.write_image("../reports/plots/sip_yoy_growth.png")
```

```python
yearly_accounts = df_04.groupby('year').agg(
    avg_active_accounts=(
        'active_sip_accounts_crore', 'mean'
    ),
    year_end_accounts=(
        'active_sip_accounts_crore', 'last'
    )
).reset_index()

yearly_accounts
```

```python
fig = px.bar(
    yearly_accounts,
    x='year',
    y='year_end_accounts',
    text='year_end_accounts',
    title='Year-end Active SIP Accounts (2022–2025)',
    labels={
        'year': 'Year',
        'year_end_accounts': 'Active SIP Accounts (Crore)'
    }
)

fig.update_traces(
    texttemplate='%{y:.2f} Cr',
    textposition='outside'
)

fig.update_layout(
    template='plotly_white',
    width=1100,
    height=600
)

fig.show()
```

```python
fig.write_image("../reports/plots/yearly_active_sip_accounts.png")
```

```python
yearly_aum = df_04.groupby('year').agg(
    avg_sip_aum=('sip_aum_lakh_crore', 'mean'),
    year_end_sip_aum=('sip_aum_lakh_crore', 'last')
).reset_index()

yearly_aum
```

```python
yearly_aum['aum_growth_pct'] = (
    yearly_aum['year_end_sip_aum'].pct_change() * 100
)

yearly_aum
```

```python
fig = px.line(
    yearly_aum,
    x='year',
    y='year_end_sip_aum',
    markers=True,
    text='year_end_sip_aum',
    title='Year-end SIP AUM Growth (2022–2025)',
    labels={
        'year': 'Year',
        'year_end_sip_aum': 'SIP AUM (₹ Lakh Crore)'
    }
)

fig.update_traces(
    texttemplate='%{y:.2f}',
    textposition='top center',
    line=dict(width=3)
)

fig.update_layout(
    template='plotly_white',
    width=1100,
    height=600
)

fig.show()
```

## Key Findings

1. Monthly SIP inflows increased throughout the analysis period, reaching a peak of ₹31,002 crore in December 2025.
2. Annual SIP inflows increased from ₹1,49,437 crore in 2022 to ₹3,35,740 crore in 2025.
3. The highest year-over-year SIP inflow growth was 46.01% in 2024.
4. Active SIP accounts increased from 6.05 crore in December 2022 to 9.35 crore in December 2025.
5. Year-end SIP AUM increased from ₹5.70 lakh crore in 2022 to ₹15.90 lakh crore in 2025.

## Conclusion

The dataset indicates an overall increase in SIP inflows, active SIP accounts, and SIP assets under management between 2022 and 2025.
