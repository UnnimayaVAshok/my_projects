```python
df = pd.read_csv("D:/My_projects/Capstone_project/data/processed/nav_history.csv")
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df = df[
    (df['date'] >= '2022-01-01') &
    (df['date'] <= '2026-12-31')
]
```

```python
print(df.isnull().sum())
```

```python
print("Number of schemes:", df['amfi_code'].nunique())
```

```python
print("Date range:", df['date'].min(), "to", df['date'].max())
```

```python
fig = px.line(
    df,
    x='date',
    y='nav',
    color='amfi_code',
    title='Daily NAV Trend Analysis (2022–2026)',
    labels={
        'date': 'Date',
        'nav': 'Net Asset Value (NAV)',
        'amfi_code': 'Scheme Code'
    },
    hover_data=['amfi_code']
)
```

```python
fig.add_vrect(
    x0='2023-01-01',
    x1='2023-12-31',
    fillcolor='green',
    opacity=0.10,
    line_width=0,
    annotation_text='2023 Bull Run',
    annotation_position='top left'
)
```

```python
fig.add_vrect(
    x0='2024-01-01',
    x1='2024-12-31',
    fillcolor='red',
    opacity=0.08,
    line_width=0,
    annotation_text='2024 Market Correction Period',
    annotation_position='top left'
)
```

```python
fig.update_layout(
    template='plotly_white',
    width=1200,
    height=650,
    hovermode='x unified',
    xaxis_title='Date',
    yaxis_title='NAV',
    legend_title='Scheme Code'
)
```

```python
fig.write_image("../reports/plots/nav_trend_analysis.png")
fig.show()
```
