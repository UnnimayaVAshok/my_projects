```python
return_cols = [
    "return_1yr_pct",
    "return_3yr_pct",
    "return_5yr_pct",
    "benchmark_3yr_pct"
]

for col in return_cols:
    Q1 = df_07[col].quantile(0.25)
    Q3 = df_07[col].quantile(0.75)
    IQR = Q3 - Q1

    anomalies = df_07[
        (df_07[col] < Q1 - 1.5 * IQR) |
        (df_07[col] > Q3 + 1.5 * IQR)
    ]

    print(f"{col}: {len(anomalies)} potential anomalies")
    print(f"\n{col}:")
    display(anomalies[["scheme_name", col]])
```

```python
expense_anomalies = df_07[
    (df_07["expense_ratio_pct"] < 0.1) |
    (df_07["expense_ratio_pct"] > 2.5)
]

print("Expense ratio anomalies:", len(expense_anomalies))
```
