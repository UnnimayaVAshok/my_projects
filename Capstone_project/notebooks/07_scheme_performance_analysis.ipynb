```python
return_cols = [
    "return_1yr_pct",
    "return_3yr_pct",
    "return_5yr_pct",
    "benchmark_3yr_pct"
]

for col in return_cols:
    Q1 = df_07[col].quantile(0.25)
    Q3 = df_07[col].quantile(0.75)
    IQR = Q3 - Q1

    anomalies = df_07[
        (df_07[col] < Q1 - 1.5 * IQR) |
        (df_07[col] > Q3 + 1.5 * IQR)
    ]

    print(f"{col}: {len(anomalies)} potential anomalies")
    print(f"\n{col}:")
    display(anomalies[["scheme_name", col]])
```

```python

df_07 = pd.read_csv("../data/processed/scheme_performance.csv")

print("Shape:", df_07.shape)

display(df_07.head())

print(df_07.info())

print(df_07.isnull().sum())
```

```python
performance_columns = [
    'return_1yr_pct',
    'return_3yr_pct',
    'return_5yr_pct',
    'benchmark_3yr_pct',
    'alpha',
    'beta',
    'sharpe_ratio',
    'sortino_ratio',
    'std_dev_ann_pct',
    'max_drawdown_pct',
    'aum_crore',
    'expense_ratio_pct'
]

df_07[performance_columns].describe().round(2)
```

```python
category_performance = (
    df_07.groupby('category')
    [['return_1yr_pct', 'return_3yr_pct', 'return_5yr_pct']]
    .mean()
    .round(2)
    .sort_values('return_3yr_pct', ascending=False)
)

display(category_performance)
```

```python
category_plot = category_performance.reset_index()

fig = px.bar(
    category_plot,
    x='category',
    y=[
        'return_1yr_pct',
        'return_3yr_pct',
        'return_5yr_pct'
    ],
    barmode='group',
    title='Mutual Fund Category-wise Returns',
    labels={
        'category': 'Fund Category',
        'value': 'Average Return (%)',
        'variable': 'Return Period'
    }
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Fund Category',
    yaxis_title='Average Return (%)',
    xaxis_tickangle=-45
)

fig.show()
```

```python
fig.write_image("../reports/plots/category_wise_returns.png")
```

```python
risk_analysis = (
    df_07.groupby('category')
    [['beta', 'sharpe_ratio', 'std_dev_ann_pct',
      'max_drawdown_pct']]
    .mean()
    .round(2)
    .sort_values('sharpe_ratio', ascending=False)
)

display(risk_analysis)
```

```python
risk_plot = risk_analysis.reset_index()

fig = px.bar(
    risk_plot.sort_values('std_dev_ann_pct', ascending=False),
    x='category',
    y='std_dev_ann_pct',
    text='std_dev_ann_pct',
    title='Annualized Volatility by Mutual Fund Category',
    labels={
        'category': 'Fund Category',
        'std_dev_ann_pct': 'Standard Deviation (%)'
    }
)

fig.update_traces(textposition='outside')

fig.update_layout(
    template='plotly_white',
    xaxis_tickangle=-45
)

fig.show()
```

```python
risk_adjusted = (
    df_07.groupby('category')
    [['sharpe_ratio', 'sortino_ratio']]
    .mean()
    .round(2)
    .sort_values('sharpe_ratio', ascending=False)
)

display(risk_adjusted)
```

```python
risk_adjusted.plot(
    kind='bar',
    figsize=(12, 6),
    width=0.8
)

plt.title('Sharpe Ratio vs Sortino Ratio by Fund Category')
plt.xlabel('Fund Category')
plt.ylabel('Risk-Adjusted Ratio')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Risk Metric')
plt.tight_layout()

plt.savefig('../reports/plots/sharpe_vs_sortino_ratio.png',
            dpi=300, bbox_inches='tight')

plt.show()
```

### Sharpe Ratio vs Sortino Ratio – Key Findings

- Liquid funds recorded the highest average Sharpe and Sortino ratios.
- Short Duration and Gilt funds also showed relatively strong risk-adjusted performance.
- ELSS funds recorded the lowest average ratios among the categories.
- Sortino ratios were higher than Sharpe ratios across all categories, indicating the difference between total volatility and downside risk.

Conclusion:

The analysis shows variations in risk-adjusted performance across mutual fund categories. However, unusually high ratios, particularly for Liquid funds, should be validated by checking the underlying calculations and data. These metrics should be considered alongside returns, volatility, and maximum drawdown rather than used independently to select funds.

```python
alpha_beta = (
    df_07.groupby('category')
    [['alpha', 'beta']]
    .mean()
    .round(2)
    .sort_values('alpha', ascending=False)
)

display(alpha_beta)
```

### Alpha and Beta Analysis – Key Findings

- Short Duration funds recorded the highest average Alpha of 1.98.
- Flexi Cap funds followed with an average Alpha of 1.82.
- Index/ETF funds recorded the highest Beta of 1.04.
- Gilt funds had the lowest Beta of 0.30.
- Most equity-oriented categories had Beta values close to 1, indicating market-like sensitivity.

Conclusion:

The analysis highlights differences in benchmark-relative performance and market sensitivity across fund categories. Short Duration funds showed the highest average Alpha, while Gilt funds had the lowest Beta. These metrics should be evaluated alongside returns, volatility, and drawdown before drawing investment conclusions.

```python
#nav_df = pd.read_csv("D:/My_projects/Capstone_project/data/processed/nav_history.csv")
cagr_data = (
    nav_df.groupby('amfi_code')
    .agg(
        beginning_nav=('nav', 'first'),
        ending_nav=('nav', 'last'),
        start_date=('date', 'first'),
        end_date=('date', 'last')
    )
)

cagr_data['years'] = (
    (cagr_data['end_date'] - cagr_data['start_date']).dt.days / 365.25
)

cagr_data['cagr_pct'] = (
    (cagr_data['ending_nav'] / cagr_data['beginning_nav'])
    ** (1 / cagr_data['years']) - 1
) * 100

display(cagr_data.head())
```

```python
nav_clean = nav_raw.copy()

nav_clean['date'] = pd.to_datetime(nav_clean['date'])

nav_clean = nav_clean[
    nav_clean['date'].notna() &
    (nav_clean['date'] <= pd.Timestamp('2026-11-06'))
].copy()

nav_clean = nav_clean.sort_values(
    ['amfi_code', 'date']
).reset_index(drop=True)

print("Cleaned records:", len(nav_clean))
print("Unique schemes:", nav_clean['amfi_code'].nunique())
print("Date range:", nav_clean['date'].min(), "to", nav_clean['date'].max())
```

```python
cagr_data = (
    nav_clean.groupby('amfi_code')
    .agg(
        beginning_nav=('nav', 'first'),
        ending_nav=('nav', 'last'),
        start_date=('date', 'first'),
        end_date=('date', 'last')
    )
)

cagr_data['years'] = (
    (cagr_data['end_date'] - cagr_data['start_date']).dt.days / 365.25
)

cagr_data['cagr_pct'] = (
    (cagr_data['ending_nav'] / cagr_data['beginning_nav'])
    ** (1 / cagr_data['years']) - 1
) * 100

display(cagr_data.head(10).round(2))
```

```python
# Highest CAGR
highest_cagr = cagr_data.sort_values(
    'cagr_pct', ascending=False
)

display(highest_cagr.head(10).round(2))
```

```python
# Lowest CAGR
lowest_cagr = cagr_data.sort_values(
    'cagr_pct', ascending=True
)

display(lowest_cagr.head(10).round(2))
```

### CAGR Analysis – Key Findings

- Calculated CAGR for 40 mutual fund schemes using their beginning and ending NAV values.
- The observation period ranged from January 2, 2022, to November 5, 2026 (approximately 4.84 years).
- DSP Small Cap Fund – Regular – Growth recorded the highest CAGR of 32.60%.
- UTI Mid Cap Fund – Regular – Growth recorded the lowest CAGR of 0.08%.
- The results show considerable variation in annualized NAV growth across schemes.
- Some schemes from the same category appeared in both the highest and lowest rankings, indicating that category alone does not determine scheme performance.

### Conclusion

CAGR helped compare the annualized NAV growth of different mutual fund schemes over a common observation period. The analysis identified schemes with relatively high and low historical NAV growth.

However, CAGR alone cannot determine investment quality. Risk measures such as Sharpe ratio, Sortino ratio, volatility, and maximum drawdown should also be considered.

**Data Limitation:** The original NAV dataset contained 27,880 records with missing dates. These records were excluded from time-based calculations, and future-dated records were also excluded. Therefore, the CAGR results are preliminary and should be interpreted with caution.
