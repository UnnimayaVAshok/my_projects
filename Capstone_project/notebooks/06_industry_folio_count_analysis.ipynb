```python
df_06['month'] = pd.to_datetime(
    df_06['month'], format='%Y-%m'
)

print("Start date:", df_06['month'].min())
print("End date:", df_06['month'].max())

print("Total records:", len(df_06))

print("\nDate intervals:")
print(df_06['month'].diff().value_counts())
```

```python
df_06 = df_06.sort_values('month').reset_index(drop=True)

df_06['folio_change'] = (
    df_06['total_folios_crore'].diff()
)

# Calculate percentage growth
df_06['folio_growth_pct'] = (
    df_06['total_folios_crore'].pct_change() * 100
)

# Display important columns
df_06[[
    'month',
    'total_folios_crore',
    'folio_change',
    'folio_growth_pct'
]]
```

```python
fig = px.line(
    df_06,
    x='month',
    y='total_folios_crore',
    markers=True,
    title='Growth of Total Mutual Fund Folios (2022–2025)',
    labels={
        'month': 'Month',
        'total_folios_crore': 'Total Folios (Crore)'
    }
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Month',
    yaxis_title='Total Folios (Crore)'
)

fig.show()
```

```python
fig.write_image("../reports/plots/total_folio_trend.png")
```

```python
# Calculate percentage contribution of each asset class
categories = [
    'equity_folios_crore',
    'debt_folios_crore',
    'hybrid_folios_crore',
    'others_folios_crore'
]

for col in categories:
    df_06[col + '_pct'] = (
        df_06[col] / df_06['total_folios_crore']
    ) * 100

# View the latest available distribution
df_06[
    ['month', 'total_folios_crore'] +
    [col + '_pct' for col in categories]
].tail()
```

```python
fig = px.area(
    df_06,
    x='month',
    y=[
        'equity_folios_crore',
        'debt_folios_crore',
        'hybrid_folios_crore',
        'others_folios_crore'
    ],
    title='Mutual Fund Folio Distribution by Asset Class',
    labels={
        'month': 'Month',
        'value': 'Folios (Crore)',
        'variable': 'Asset Class'
    }
)

fig.update_layout(template='plotly_white')

fig.show()
```

```python
fig.write_image("../reports/plots/folio_distribution_by_asset_class.png")
```

```python
asset_class_growth = pd.DataFrame({
    'Asset Class': ['Equity', 'Debt', 'Hybrid', 'Others'],
    'Start Folios': [
        df_06['equity_folios_crore'].iloc[0],
        df_06['debt_folios_crore'].iloc[0],
        df_06['hybrid_folios_crore'].iloc[0],
        df_06['others_folios_crore'].iloc[0]
    ],
    'End Folios': [
        df_06['equity_folios_crore'].iloc[-1],
        df_06['debt_folios_crore'].iloc[-1],
        df_06['hybrid_folios_crore'].iloc[-1],
        df_06['others_folios_crore'].iloc[-1]
    ]
})

asset_class_growth['Absolute Growth'] = (
    asset_class_growth['End Folios'] -
    asset_class_growth['Start Folios']
)

asset_class_growth['Growth %'] = (
    asset_class_growth['Absolute Growth'] /
    asset_class_growth['Start Folios']
) * 100

asset_class_growth
```

```python
fig = px.bar(
    asset_class_growth,
    x='Asset Class',
    y='Growth %',
    text='Growth %',
    title='Folio Growth by Asset Class (2022–2025)'
)

fig.update_traces(texttemplate='%{text:.2f}%', textposition='outside')

fig.update_layout(
    template='plotly_white',
    yaxis_title='Growth (%)'
)

fig.show()
```

```python
fig.write_image("../reports/plots/asset_class_folio_growth.png")
```

## Key Findings

1. Total mutual fund folios increased from 13.26 crore in January 2022 to 26.12 crore in December 2025.
2. Equity accounted for approximately 70% of total folios.
3. Equity recorded the highest absolute growth of 9 crore folios.
4. All four asset classes recorded approximately 96% growth.

## Conclusion

The analysis indicates an overall increase in mutual fund folio counts during the observed period. Equity remained the dominant asset class by folio share.
